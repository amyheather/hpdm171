### 4.4. Advantages of returning values from a function

* If a function returns a value you can pass it to another function.

```python
def add(num1, num2):
    return num1 + num2

def square(num):
    return num ** 2  
```

```python
result = square(add(1, 1))
print(result)  
```

## 4.6 In Python Functions can return multiple values

```python
def list_info(data):
    '''
    Returns Sum, Length and Mean of a list of integers
    
    Parameters:
    ----------
    data: list
        a Python list containing numeric data

    Returns:
    --------
    tuple(int, int, float)
    
    '''
    list_sum = sum(data)
    list_length = len(data)
    list_mean = list_sum / list_length
    
    return list_sum, list_length, list_mean
```

```python
data = [1, 2, 3, 4, 5]

results = list_info(data)
print(f"The variable 'results': {results} has type {type(results)}")

data_sum, data_length, data_mean = list_info(data)
print(f'Seperate variables: sum {data_sum}, length {data_length}, mean {data_mean}')
```
