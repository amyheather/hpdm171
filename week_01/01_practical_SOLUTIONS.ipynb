## 6.4 Test your knowledge: Discounting future QALYs

In health economics we often want to compare an intervention that provides health benefits in the future with one that provides benefits now. To do this, we **discount** future health gains measured in Quality Adjusted Life Years (QALYs) back to their **present value**, using a discount rate $r$.

For a single lump of QALYs received \(n\) years in the future, the present value is:

$$\text{PV QALYs} = \frac{\text{Future QALYs}}{(1 + r)^n}$$

**Tasks:**

1. Write a Python function called `qaly_pv` that calculates the present value of future QALYs using the formula above.  
   - The function should take three parameters:  
     - `future_qalys` (float): QALYs gained in the future  
     - `discount_rate` (float): annual discount rate $r$ (e.g. `0.035` for 3.5%)  
     - `years_ahead` (float): number of years in the future $n$
2. Use your function to calculate the present value of **5 QALYs** gained **10 years** from now at a discount rate of **3.5% per year** (i.e. `discount_rate = 0.035`).
3. **Return** the result to the calling code.
4. Print the result, rounding to 3 decimal places.

````python
def qaly_pv(future_qalys, discount_rate, years_ahead):
    """
    Discount future QALYs back to present value.

    Parameters
    ----------
    future_qalys : float
        QALYs gained in the future.
    discount_rate : float
        Annual discount rate (e.g. 0.035 for 3.5%).
    years_ahead : float
        Years in the future.

    Returns
    -------
    float
        Present value of future QALYs.
    """
    return future_qalys / (1 + discount_rate) ** years_ahead
````

````python
# use this code to test your function
discounted_qaly = qaly_pv(5, 0.035, 10)
print(f"Discounted QALY: {discounted_qaly:.3f}")
````
