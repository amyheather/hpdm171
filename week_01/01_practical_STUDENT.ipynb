In this exercise we will learn:

### 3.1. Why do we use variables?

We want to be able to store data and results of calculations in ways we can re-use. For this we define **variable names**.

* Variable names can only contain letters, numbers, and underscores. A name must start with a letter or an underscore `_`;
* **Spaces are not allowed** in variable names, so we use underscores instead of spaces. For example, use `student_name` instead of `student name`
* Should not include a hyphen `-`, which can be interpreted as a minus sign
* Variable names should be descriptive, without being too long. For example `n_patients` is better than just `patients`, and `number_of_patients_in_the_hospital`.
* A descriptive name can also help with code readability.

Some other examples of how to improve naming include:

| Bad variable name                                     | Better variable name |
| ----------------------------------------------------- | -------------------- |
| x                                                     | n_patients           |
| temp                                                  | temperature_c        |
| data1                                                 | patient_data         |
| number_of_patients_in_the_hospital_today_including_ED | n_patients           |
| a1                                                    | mean_age             |
| myvar                                                 | admission_rate       |


Here is an example of storing data in a integer variable called `n_patients`

````python
n_patients = 30_000
print(n_patients)
print(f'The number of patients in the emergency department is {n_patients}')
````

### 3.2 What is a variable?

A variable name is a **label** that points to a value stored in your computer's memory. When you write `n_patients = 30_000`, you are **assigning** the integer value `30_000` to the variable `n_patients `and attaching that label to this value.

In the example above, you can think of the variable as a post‑it note with `n_patients` written on it. This label points to the integer value `30_000` in memory. When we ask Python to `print(n_patients)`, it looks up the value attached to that label and displays `30_000`.

In Python we can also move this label to point at a different value (even of a different type). In the example below, `number` is first assigned the value `1.2`, and then re-assigned a text value `"One point two"`. Python updates the label each time you assign a new value.

````python
number = 1.2
print(number)
number = "One point two"
print(number)
````

### 3.3 Mathematics and variables

Variables means that we can express mathematical operations using easy to read syntax.

For example, let's assume a patient has body mass index (BMI) of 32.0 and a weight loss intervention has been shown to reduce bmi by 10%.  We can write out that computation in Python using variable names rather than the quantitative values themselves.

````python
bmi = 32.0
reduction_factor = 0.1  # e.g. 10% BMI reduction after an intervention

bmi_after_intervention = bmi * (1 - reduction_factor)

print(bmi_after_intervention)
````

## 4. Comments in Python

In this exercise we will learn how to

* ✅ Add comments to our code using the `#` symbol
* ✅ Write comments that explain the why, not just repeat the code
* ✅ Use comments to document more complex code clearly

### 4.1 Adding comments

Comments allow you to write in you native language (e.g. English), within your program. In Python, any line that starts with a hash `#` symbol is ignored by the Python interpreter.

Typically put a space between `#` and a comment. For example:

````python
# Commented out text
print("This line is not a comment, it is code.")

print("Python will ignore comments")  # this is an inline comment 
````

### 4.2 What makes a good comment?

* It is short and to the point.
* It explains the **why** (your reasoning or intent), not just the **what** the code already makes obvious.
* It helps future‑you and other people quickly understand the purpose of the code or any non‑obvious decisions.
* It focuses on clarifying tricky or non‑standard parts of the code, rather than narrating every line.

> If a comment just repeats what the code says, it is usually not helpful and may even become misleading if the code changes later.  In practical terms this means don't just comment everything if it is self evident!

### 4.3 An example of commenting BMI code

In the code below, the comments focus on explaining why we are doing particular calculations, not just restating what each line does.

* The comment above the BMI calculation explains the formula being implemented, which might not be obvious from the code alone.
* The comment above the high_risk line explains the rule we are using to flag a high‑risk patient (high BMI and older than a certain age).
* We do not add comments to every assignment like weight_kg = 82, because the variable names already make their purpose clear.
* Good comments should clarify intent and non‑obvious logic, rather than narrate each line of code.

````python
weight_kg = 82
height_m = 1.75
age_years = 68

# Calculate Body Mass Index (BMI) = weight (kg) / height^2 (m^2)
bmi = weight_kg / (height_m ** 2)

bmi_risk_threshold = 30
age_risk_threshold = 65

# high‑risk flag: high BMI AND older than threshold
high_risk = (bmi >= bmi_risk_threshold) and (age_years >= age_risk_threshold)

print("BMI:", bmi)
print("High risk patient:", high_risk)
````

### 5.1 Creating lists

A Python `List` is a simple and flexible way to store lots of variables (of any type of data).

````python
foo = [0, 1, 2, 3]
foo
````

The square brackets `[]` say that what follows will be a list: a collection of objects. The commas separate the different objects contained within the list.  We assigned the list to a variable called `foo`

In Python, a list can hold *anything*. For example:

````python
bar = [0, 1.2, "hello", [3, 4]]
bar
````

### 5.2 List items

The contents of a list are referred to as **items**. To access individual items in a list, use square brackets again. The items are ordered left-to-right, and the first item has **index** `0`

> Lists are **zero indexed**. This can be confusing at first, but its very important to remember.

````python
print(foo[0])
print(foo[3])
print(bar[1])
print(bar[3])
````

> **Note** If we try to access an element that isn't in the list we get an error.

### 5.3 Assigning items

We can assign the value of elements of a list in the same way as any variable:

````python
foo[1] = 999
foo
````

````python
foo[3] = -1
foo
````

### 5.4 List slicing

The number in square brackets is called the **index** of the item. Because lists start at zero, the index of an item is always one less than its position in the list. So to get the second item in the list, we need to use an index of 1.

We can work with multiple elements of a list at once using *slicing* notation. Here are some examples:

````python
print(foo)
print(foo[0:2])
print(foo[:2])
print(foo[1:])
print(foo[0:4:2])
print(foo[::2])
````

The notation `[start:end:step]` means to return the entries from the `start`, up to **but not including** the `end`, in steps of length `step`. If the `start` is not included (e.g. `[:2]`) it defaults to the start, i.e. `0`. If the `end` is not included (e.g. `[1:]`) it defaults to the end (i.e., `len(...)`). If the `step` is not included it defaults to `1`.

To get the last item in a list, no matter how long the list is, you can use an index of -1. This syntax also works for the second to last item, the third to last, and so forth. You can't use a negative number larger than the length of the list, however.

````python
print(foo)
print(foo[-1]) # last item 
print(foo[-2]) # 2nd to last item
````

### 5.5. Searching a list

If you want to find out the position of an element in a list, you can use the `index()` method. This method returns a `ValueError` if the requested item is not in the list.

````python
foo.index(999)
````

You can test whether an item is in a list using the "in" keyword. This will become more useful after learning how to use if-else statements.

````python
999 in foo
````

````python
11 in foo
````

### 5.6 Appending, inserting, and deleting list items

We can add an item to the end of a `list` using the `append()` method. This method adds the new item to the end of the list.

````python
foo.append(12)
print(foo)
````

We can also insert items anywhere we want in a list, using the `insert()` method. We specify the position we want the item to have, and everything from that point on is shifted one position to the right. In other words, the index of every item after the new item is increased by one.

````python
foo.insert(1,9999)
print(foo)
````

We can remove an item from a list using the `del` statement.  You need to specify the index you wish to remove.

````python
del foo[2]
print(foo)
````

### 6.3 Defining Custom Python Functions

So far we have used functions built-in to python such as `print()` and `math.cos()`

In real health data science applications, you will need to define your own custom functions in Python.  Here we create a toy example of calculating a risk score for a patient at risk of a stroke.  

````python
def total_risk_score(stroke_risk_score, bleeding_risk_score):
    """
    Returns a simulated combined risk score from two components.

    Parameters
    ----------
    stroke_risk_score : float
        Estimated risk of stroke
    bleeding_risk_score : float
        Estimated risk of bleeding

    Returns
    -------
    float
        Total combined risk score.
    """
    return stroke_risk_score + bleeding_risk_score
````

Let's break down the function. First of all let's take a look the function definition, name and signature:

```python
def total_risk_score(stroke_risk_score, bleeding_risk_score):
    ...
```

The keyword used to **define** a new function in Python is `def`. You write `def`, followed by the function name (here `total_risk_score`), then a pair of round brackets `()`, and then a colon `:` at the end of the line.

Inside the brackets you list the **parameters** the function expects, separated by commas. These are *names* you choose that will be used inside the function to refer to whatever values the user passes in. For example, when someone calls `total_risk_score(3.0, 2.0)`, the first argument `3.0` is available inside the function as `stroke_risk_score`, and the second as `bleeding_risk_score`.

The colon at the end of the `def` line tells Python that the following indented lines form the **body** of the function: 

```python
return stroke_risk_score + bleeding_risk_score
```

This is the code that will run whenever the function is called. Every line that belongs to the function must be indented by the same amount (conventionally four spaces). In JupyterLab, this indentation will usually be added for you automatically after you press Enter.  The `return` keyword tells Python what value to return to the calling code.   

The triple-quoted string immediately under the `def` line is called a **docstring**. 

```python
    """
    Returns a simulated combined risk score from two components.

    Parameters
    ----------
    stroke_risk_score : float
        Estimated risk of stroke
    bleeding_risk_score : float
        Estimated risk of bleeding

    Returns
    -------
    float
        Total combined risk score.
    """
```

It does not change how the function runs, but it serves as documentation: it explains what the function does, what its parameters mean, and what it returns. In practice, a function without a useful docstring is “broken” from a readability and maintenance point of view, because other people (and future you) cannot easily understand how to use it. You can view a function’s docstring by calling `help(total_risk_score)` or by using the `?total_risk_score` syntax in a notebook.

````python
help(total_risk_score)
````

* Let's call our new function

````python
stroke_risk = 3.0
bleeding_risk = 2.0
total = total_risk_score(stroke_risk, bleeding_risk)
print(f"Total combined risk score is {total}")
````
