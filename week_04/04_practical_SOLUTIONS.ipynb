## Exercise 1: Oxygen saturation monitoring (`while` loop)

```python
spo2 = 88
hour = 0
target_spo2 = 94

while spo2 < target_spo2:
    print(f"Hour {hour}: oxygen saturation is {spo2}%")
    spo2 += 2
    hour += 1

print(f"Target reached at hour {hour}: oxygen saturation is {spo2}%")
```

## Exercise 2: Counting abnormal heart rates

```python
def heart_rate_in_normal_range(heart_rate_bpm):
    """Return True if resting heart rate is within the normal range."""
    return heart_rate_bpm >= 60 and heart_rate_bpm <= 100
```

```python
def count_abnormal_heart_rates(heart_rates):
    """Return a count of heart rates outside the normal resting range."""
    count = 0
    for heart_rate in heart_rates:
        if not heart_rate_in_normal_range(heart_rate):
            count += 1
    return count
```

```python
heart_rates = [72, 110, 58, 95, 102, 67]
print(count_abnormal_heart_rates(heart_rates))  # expected: 3
```

## Exercise 4: Update and query a patient record (dictionaries)

```python
def add_observations(patient_record, observations):
    """Add a dictionary of observations to a patient record and return it."""
    patient_record.update(observations)
    return patient_record
```

```python
def get_observation(patient_record, key, default="not recorded"):
    """Safely retrieve a value from a patient record."""
    return patient_record.get(key, default)
```

```python
patient = {"patient_id": 1003, "age": 45, "heart_rate": 88, "ward": "C1"}

new_obs = {"heart_rate": 95, "oxygen_saturation": 98, "temperature": 37.2}
patient = add_observations(patient, new_obs)
print(patient)

print(get_observation(patient, "temperature"))     # expected: 37.2
print(get_observation(patient, "blood_pressure"))  # expected: not recorded
print(get_observation(patient, "ward"))             # expected: C1
```

## Exercise 5: Debugging a scope problem (new exercise)

**Task 1 — explanation:**

`is_high_heart_rate` does not take `threshold` as a parameter. Instead it reads `threshold` directly from notebook-level (global) scope. The first call reads `threshold = 100`, so `110 > 100` is `True`. Before the second call, the notebook-level variable `threshold` is reassigned to `130`. Because the function looks up `threshold` fresh from the notebook's global scope every time it runs, the second call evaluates `110 > 130`, which is `False`. The function's behaviour silently changed even though its argument did not — this is exactly the bad practice described in Section 7 of the consolidation lecture (functions should not rely on notebook-level variables).


```python
# Task 2: fixed function — threshold is now a parameter with a sensible default
def is_high_heart_rate_fixed(heart_rate_bpm, threshold=100):
    """Return True if heart rate is above threshold (default 100 bpm)."""
    return heart_rate_bpm > threshold
```

```python
# Task 3: demonstrate the fixed function is unaffected by a global 'threshold'
threshold = 999  # this should have NO effect on the function below

print(is_high_heart_rate_fixed(110))                 # True  - uses default threshold=100
print(is_high_heart_rate_fixed(110, threshold=130))  # False - caller explicitly overrides
```

## Exercise 6: Putting it together — ward review summary

```python
def attendance_flag(days_since_last_attend, n_attends_last_year):
    """Return an attendance flag based on recency and frequency of ED attendance."""
    is_recent = days_since_last_attend <= 7
    is_frequent = n_attends_last_year >= 5
    if is_recent and is_frequent:
        return "frequent recent attender"
    return "routine"
```

```python
patients = [
    {"name": "Asha", "heart_rate": 72, "days_since_last_attend": 3, "n_attends_last_year": 6},
    {"name": "Ben", "heart_rate": 110, "days_since_last_attend": 30, "n_attends_last_year": 1},
    {"name": "Chloe", "heart_rate": 58, "days_since_last_attend": 3, "n_attends_last_year": 1},
    {"name": "David", "heart_rate": 88, "days_since_last_attend": 2, "n_attends_last_year": 7},
]

summaries = []

for patient in patients:
    hr_status = "normal" if heart_rate_in_normal_range(patient["heart_rate"]) else "abnormal"
    flag = attendance_flag(patient["days_since_last_attend"], patient["n_attends_last_year"])
    summary = f"{patient['name']}: heart rate {hr_status}, {flag}"
    summaries.append(summary)

for line in summaries:
    print(line)
```
