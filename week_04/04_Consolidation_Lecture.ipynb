## 1. Warm up: why do we loop at all?

Recall from Week 2 the motivating example: a healthcare assistant recording the same observation for several patients. Repeating the same three steps manually for each patient does not scale. A **loop** lets us describe the repeated behaviour once.


```python
patients = ["Phil", "Sarah", "Becky", "Reggie"]

for patient_name in patients:
    print(f"Recording blood pressure for {patient_name}")
```

## 2. `for` loops versus `while` loops (recap)

* Use a `for` loop when you know the number of iterations **in advance** (e.g. looping over a fixed list of patients).
* Use a `while` loop when you do **not** know the number of iterations in advance, and instead loop until some condition becomes `True` (or `False`).
* Both loop types require correct **indentation** after the `:`
* Forgetting to include indentation error raises an `IndentationError`.


### 2.1 `for` loop with `range`

```python
# range(start, stop) is 'up to and excluding' stop
for patient_num in range(1, 4):
    print(f"Go to Patient {patient_num}")
    print(f"Record blood pressure for Patient {patient_num}")
```

### 2.2 `while` loop: monitoring until a target is reached

```python
spo2 = 88
hour = 0
target_spo2 = 94

while spo2 < target_spo2:
    print(f"Hour {hour}: oxygen saturation is {spo2}%")
    spo2 += 2
    hour += 1

print(f"Target reached at hour {hour}: oxygen saturation is {spo2}%")
```

> **Common mistake:** forgetting to update the variable being tested inside a `while` loop causes an **infinite loop**. Always check that something inside the loop body moves you closer to the exit condition.


## 3. Looping + conditionals together (recap)

Loops are often combined with `if` / `elif` / `else` logic. A classic teaching example is FizzBuzz, which I have recast here as a ward-round task allocator.


```python
def ward_round_task(patient_number):
    """
    Return a ward-round task string based on patient_number.

    Parameters
    ----------
    patient_number : int

    Returns
    -------
    str
    """
    if patient_number % 3 == 0 and patient_number % 5 == 0:
        return "blood pressure check + medication review"
    elif patient_number % 3 == 0:
        return "blood pressure check"
    elif patient_number % 5 == 0:
        return "medication review"
    else:
        return "routine review"
```

```python
limit = 15
results = []

for patient_i in range(1, limit + 1):
    results.append(ward_round_task(patient_i))

results
```

## 4. Nested loops (recap)

A nested loop is a 'loop of loops'. For every single iteration of the **outer** loop, the **inner** loop runs through all of its iterations.


```python
# Build a simple triangular pattern using an outer and inner loop
for outer_index in range(1, 6):
    for inner_index in range(1, outer_index + 1):
        print(inner_index, end='')
    print('')
```

## 5. Looping over lists: `enumerate()` and `zip()` (recap)

* Loop directly over a list when you only need the **values**.
* Use `enumerate()` when you need both the **index/position** and the **value**.
* Use `zip()` to loop over two (or more) lists **together**, pairing up corresponding elements.
* Always check that lists passed to `zip()` are the same length — `zip()` silently truncates to the shortest list.


```python
patients = ['Phil', 'Sarah', 'Becky', 'Reggie']
results = [10.0, 2.5, 25.1, 35.0]

if len(patients) != len(results):
    print("Error, lists are different lengths")
else:
    for patient_id, (patient_name, result_i) in enumerate(zip(patients, results), start=1):
        print(f"Patient {patient_id} ({patient_name}): {result_i}")
```

## 6. Functions revisited: parameters, defaults, and docstrings (recap)

Functions let us reuse the same logic without repeating code, and they combine naturally with loops: call the function once per loop iteration.


```python
def calculate_bmi(weight_kg, height_m, decimals=1):
    """
    Calculate body mass index (BMI), rounded to a chosen number of decimal places.

    Parameters
    ----------
    weight_kg : int or float
        Patient weight in kilograms.
    height_m : int or float
        Patient height in metres.
    decimals : int, optional
        Number of decimal places to round to (default is 1).

    Returns
    -------
    float
        Body mass index.

    Raises
    ------
    ValueError
        If weight_kg <= 0 or height_m <= 0.
    """
    if weight_kg <= 0:
        raise ValueError("weight_kg must be greater than 0")
    if height_m <= 0:
        raise ValueError("height_m must be greater than 0")

    bmi = weight_kg / (height_m ** 2)
    return round(bmi, decimals)
```

### 6.1 Combining a function with a loop

> Notice that each item in `patient_measurements` below is a `tuple`, not a `list` (Week 1). We are not planning to modify an individual patient's data in place, so an immutable tuple is a sensible, lightweight choice here and the `for` loop unpacks each tuple into three named variables in one line.


```python
patient_measurements = [
    ("Phil", 82, 1.78),
    ("Sarah", 70, 1.80),
    ("Becky", 58, 1.65),
]

for name, weight_kg, height_m in patient_measurements:
    bmi = calculate_bmi(weight_kg, height_m)
    print(f"{name}: BMI = {bmi}")
```

## 7. Variable and function scope (recap)

Now that we are writing our own functions with parameters, it is worth revisiting **scope**: the part of a program where a variable can be seen and used.

* A **notebook-level** (global) variable is created in a notebook cell and is available in any cell run afterwards.
* A **function-level** (local) variable is created inside a function body and is normally only available **inside that function** . It disappears once the function finishes running.
* A function *can* read a notebook-level variable, but this is **bad practice**: it makes the function harder to reuse and can cause hidden bugs if cells are run out of order. Prefer passing everything a function needs in as a **parameter**.


### 7.1 Bad practice: relying on a notebook-level variable

```python
# Notebook-level (global) variable
clinic_name = "Exeter Community Clinic"

def attendance_message(patient_name):
    """Return an attendance message using the notebook-level clinic name."""
    # BAD PRACTICE: this function relies on a variable defined outside itself
    return f"{patient_name} attended {clinic_name}"

print(attendance_message("Amber"))
```

### 7.2 Best practice: pass everything in as a parameter

```python
def attendance_message_good(patient_name, clinic):
    """Return an attendance message using only function parameters."""
    return f"{patient_name} attended {clinic}"

# The function is now self-contained: it works for any clinic, not just clinic_name
print(attendance_message_good("Amber", "Exeter Community Clinic"))
print(attendance_message_good("David", "Royal Devon University Hospital"))
```

### 7.3 A common source of confusion

If you assign to a variable inside a function, Python treats it as a **new local variable** unless told otherwise. It does not overwrite a notebook-level variable of the same name.

```python
status = "not reviewed"  # notebook-level variable

def review_patient():
    """Print a local review status to demonstrate function scope."""
    status = "reviewed"  # this is a different, local variable
    print("Inside function:", status)

print("Before function call:", status)
review_patient()
print("After function call:", status)
```

> **Common mistake:** expecting a variable created inside a function to be visible afterwards in the notebook, or expecting a function to automatically "see" an updated notebook-level variable. If your function needs a value, pass it in as a parameter; if you need a result back, `return` it.


## 8. Dictionaries + loops (recap)

Dictionaries store key–value pairs, which is useful when each patient has several pieces of information. Looping over `.items()` lets us process every field.


```python
patient_record = {
    "id": 1001,
    "age": 67,
    "heart_rate": 88,
    "ward": "A3",
}

for key, value in patient_record.items():
    print(f"{key}: {value}")
```

## 9. Mini case study: bringing it all together

**Task:** for a small set of patients, calculate BMI, classify resting heart rate, and produce a one-line summary for each patient.
To do this you are combining a function, a loop, conditional logic, scope (each function is self-contained and only uses its own parameters), and a dictionary.


```python
def heart_rate_category(heart_rate_bpm):
    """Return a simple category for a resting heart rate in bpm."""
    if heart_rate_bpm < 60:
        return "low"
    elif heart_rate_bpm <= 100:
        return "normal"
    else:
        return "high"
```

```python
patients = [
    {"name": "Phil", "weight_kg": 82, "height_m": 1.78, "heart_rate": 72},
    {"name": "Sarah", "weight_kg": 70, "height_m": 1.80, "heart_rate": 130},
    {"name": "Becky", "weight_kg": 58, "height_m": 1.65, "heart_rate": 55},
]
```

```python
summaries = []

for patient in patients:
    bmi = calculate_bmi(patient["weight_kg"], patient["height_m"])
    category = heart_rate_category(patient["heart_rate"])
    summary = f"{patient['name']}: BMI {bmi}, heart rate {patient['heart_rate']} bpm ({category})"
    summaries.append(summary)

for line in summaries:
    print(line)
```
