## 1. Boolean Operators 

* We have seen that Python has a type `bool` that can take two values `True` or `False`
* Boolean operators return `True` and `False` (or yes/no) answers


| Operation       | Description       | Operation       | Description |
| :------------- | :----------: | -----------: | -----------:|
| `a == b`    | `a` equal to `b`              | `a != b`    | `a` not equal to `b`             |
| `a < b`     | `a` less than `b`             | `a > b`     | `a` greater than `b`             |
| `a <= b`    | `a` less than or equal to `b` | `a >= b`    | `a` greater than or equal to `b` |


* **Tip**: Remember that the `=` operator is for **assignment** 
* use `==` when you want to **compare** equality

````python
age = 25
print(age < 40)
print(age >= 40)
print(age == 25)
````

## 2. Conditional Logic

* Most code makes extensive use of **if then** statements
* This makes use of the `if`, `elif`, and `else` keywords   

```python
if <Boolean Operation 1>:
    do something
elif <Boolean Operation 2>:
    do something different
else:
    take default action
``` 

### 2.1 A simple example of an `if` then `else` statement
* Is the patient classed as an an adult or a child?

````python
patient_age = 5

if patient_age >= 18:
    print("Patient is classified as an adult")
else:
    print("Patient is classified as a child")
````

### 2.2 A more complex **if then**

````python
# Define safe resting heart rate range (beats per minute)
lower_limit = 60
upper_limit = 100

# Measured resting heart rate for a patient
heart_rate = 110

if heart_rate >= lower_limit and heart_rate <= upper_limit:
    print('Heart rate is within the normal resting range')
else:
    print('Heart rate is outside the normal resting range')
````

### 2.3 Else if = `elif`

We may have multiple if statements chained together. 

````python
bmi = 32.0

# Python checks each condition in order and stops at the first one that is True
if bmi < 0:
    print('Error. Please enter a BMI greater than zero')
elif bmi < 18.5:
    print('BMI category: Underweight')
elif bmi < 25:
    print('BMI category: Normal weight')
elif bmi < 30:
    print('BMI category: Overweight')
else:
    # If none of the above conditions are True, BMI must be >= 30
    print('BMI category: Obese')
````

## 3. Functions and if statements

* We could also use a function within an `if` statement

````python
def heart_rate_in_normal_range(heart_rate_bpm):
    """
    Return True or False to indicate if
    resting heart rate is within the normal range.

    Parameters
    ----------
    heart_rate_bpm : int or float
        Resting heart rate in beats per minute.

    Returns
    -------
    bool
    """
    # Normal resting heart rate range: 60–100 bpm
    return heart_rate_bpm >= 60 and heart_rate_bpm <= 100
````

````python
heart_rate = 110

if heart_rate_in_normal_range(heart_rate):
    print('Heart rate is within the normal resting range')
else:
    print('Heart rate is outside the normal resting range')
````

* We can also use the `not` operator in Python

````python
heart_rate = 110

if not heart_rate_in_normal_range(heart_rate):
    # this branch is now executed if function returns False
    print('Heart rate is outside the normal resting range')
else:
    print('Heart rate is within the normal resting range')
````

## 4. Nested if statements

* To help with complex logic you can nest if statments inside each other.

````python
def ed_reattendance_risk_score(days_since_last_attend, n_attends_last_year):
    """
    Toy example: return a simple numeric risk score for ED re-attendance.

    Higher score if:
    - The patient attended very recently (within 7 days), and
    - They are a frequent attender (>= 5 visits in the last year).

    Parameters
    ----------
    days_since_last_attend : int
        Days since the patient's last ED attendance.
        Use a large number (e.g. 999) if no prior attendance on record.
    n_attends_last_year : int
        Number of ED attendances in the last 12 months.

    Returns
    -------
    float
        Re-attendance risk points (toy score for teaching).
    """
    if days_since_last_attend <= 7:
        # Recently attended – now check how frequent an attender they are
        if n_attends_last_year >= 5:
            return 2.0  # recent + frequent
        else:
            return 1.0  # recent but not a frequent attender
    else:
        # Not a recent attendance – separate rule based on frequency alone
        if n_attends_last_year >= 5:
            return 1.0  # frequent but not recent
        else:
            return 0.0  # neither recent nor frequent
````

````python
days_since_last_attendance = 3
n_attendances_last_year = 6

risk_score = ed_reattendance_risk_score(days_since_last_attendance,
                                        n_attendances_last_year)

print(f"Patients risk score is {risk_score}")
````

## 5. An alternative to nested conditionals

* We could create boolean flags first and then have a simpler `if then` structure.

````python
def ed_reattendance_risk_score(days_since_last_attend, n_attends_last_year):
    """
    Toy example: return a simple numeric risk score for ED re-attendance
    based on recency and frequency of past visits.
    """
    # Use boolean expressions to store True/False "flags"
    is_recent = days_since_last_attend <= 7
    is_frequent = n_attends_last_year >= 5

    # Then look up the risk score based on the combination of flags
    if is_recent and is_frequent:
        return 2.0          # recent + frequent
    elif is_recent and not is_frequent:
        return 1.0          # recent only
    elif (not is_recent) and is_frequent:
        return 1.0          # frequent only
    else:
        return 0.0          # neither recent nor frequent
````

````python
days_since_last_attendance = 3
n_attendances_last_year = 6

risk_score = ed_reattendance_risk_score(days_since_last_attendance,
                                        n_attendances_last_year)

print(f"Patients risk score is {risk_score}")
````

## Introduction to iterating over data using **loops**

**Algorithms often need to do the same thing again and again**

For example, an algorithm making three servings of toast

**Making Toast Algorithm**:

1. Put a slice of bread in the toaster
2. Push lever down to turn on the toaster
3. Remove the toasted bread from the toaster
4. Put a slice of bread in the toaster
5. Push lever down to turn on the toaster
6. Remove the toasted bread from the toaster
7. Put a slice of bread in the toaster
8. Push lever down to turn on the toaster
9. Remove the toasted bread from the toaster


**A better way to do this in code is to use a LOOP**

Do the following 3 steps, 3 times:
1. Put a slice of bread in the toaster
2. Push lever down to turn on the toaster
3. Remove the toasted bread from the toaster



* There are two types of loop in Python
* `for` loops and `while` loops
* We generally use `while` if we do not know the number of iterations in advance
* We generally use `for` if we know the number of iterations in advance

### While loops

````python
age = 15

while age <= 18:
    print(f'you are currently {age} years old')
    age += 1
````

### While Loop Structure

* All while loops have the same structure
* You use the `while` keyword 
* Followed by a boolean operation (e.g. `age <= 18`)
* Beware of infinite loops!
* Let's test a more complex while loop using a function

````python
def fizzbuzz(n):
    """
    For multiples of three print "Fizz" instead of the number 
    and for the multiples of five print "Buzz". 
    For numbers which are multiples of both three 
    and five print "FizzBuzz".
    
    Parameters:
    ----------
    n: int
        the number to test
    """
    if n % 3 == 0 and n % 5 == 0:
        print('fizzbuzz')
    elif n % 3 == 0:
        print('fizz')
    elif n % 5 == 0:
        print('buzz')
    else:
        print(n)
````

````python
n = 1
limit = 15

while n <= limit:
    fizzbuzz(n)
    n += 1
````

* A `while` loop example where we do **not** know the number of iterations in advance.

````python
list_to_search = ['we', 'are', 'the', 'knights', 'who', 'say', 'ni']
not_found = True
word_to_find = 'knights'
current_index = 0
index_of_word = -1

while not_found:
    if list_to_search[current_index] == word_to_find:
        not_found = False
        index_of_word = current_index
    current_index += 1

print(f"the word '{word_to_find}' is located in index {index_of_word}")
````

### For loops

* To create a **for** loop you need the following:
* `for` keyword
* a variable 
* the `in` keyword
* the `range()` function - which is a built-in function in the Python library to create a sequence of numbers


````python
for age in range(5):
    print(f'you are currently {age} years old')
````

* notice that the loop sets age to 0 initially!
* `range()` takes keyword arguments to set the start (inclusive, default = 0), end (exclusive) and step

````python
for age in range(1, 5):
    print(f'you are currently {age} years old')
````

````python
for age in range(1, 5, 2):
    print(f'you are currently {age} years old')
````

## Nested Loops
* `for` and `while` loops can be nested within each other.
* Think of nested loops as a 'loop of loops'
* Remember that for each iteration outer loop will consist of multiple iterations of an inner loop
* **Don't panic** if you do not understand straight away!

````python
for outer_index in range(3):
    print(f'Outer loop iteration: {outer_index}')
    
    for inner_index in range(5):
        print(f'\tInner loop iteration: {inner_index}')
        
````

* Example 2: The inner loop now iteratures **backwards**

````python
for outer_index in range(2):
    print(f'Outer loop iteration: {outer_index}')
    
    for inner_index in range(5, 0, -1):
        print(f'\tInner loop iteration: {inner_index}')
````

Nested Loops Example 2.
* Let's use a nested `for` loop to create the pattern below


```
1

123
1234
12345
```

````python
for outer_index in range(1, 6):
    #remember this is a loop of loops. 
    #the loop below execute all iterations each 
    # time the outer loop iterates
    for inner_index in range(1, outer_index + 1):
        
        #we use the end='' option of print so 
        #that it prints on the same line as previous
        print(inner_index, end='')
    
    #new line
    print('')
        
````
