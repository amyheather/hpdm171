## Exercise 1: BMI category


```python
def bmi_category(bmi):
    """
    Return a BMI category.

    Parameters
    ----------
    bmi : float

    Returns
    -------
    str
    """
    if bmi < 0:
        return "error"
    elif bmi < 18.5:
        return "underweight"
    elif bmi < 25:
        return "normal weight"
    elif bmi < 30:
        return "overweight"
    else:
        return "obese"
```

```python
# test data
print(bmi_category(17.2))
print(bmi_category(22.8))
print(bmi_category(28.1))
print(bmi_category(31.0))
```

## Exercise 2: ED reattendance risk score


```python
def ed_reattendance_risk_score(days_since_last_attend, n_attends_last_year):
    """
    Return a simple risk score simulating the risk that
    a patient will re-attend the emergency department.

    Parameters
    ----------
    days_since_last_attend : int
    n_attends_last_year : int

    Returns
    -------
    float
    """
    if days_since_last_attend <= 7 and n_attends_last_year >= 5:
        return 2.0
    elif days_since_last_attend <= 7 and n_attends_last_year < 5:
        return 1.0
    elif days_since_last_attend > 7 and n_attends_last_year >= 5:
        return 1.0
    else:
        return 0.0
```

```python
print(ed_reattendance_risk_score(3, 6))
print(ed_reattendance_risk_score(3, 1))
print(ed_reattendance_risk_score(30, 6))
print(ed_reattendance_risk_score(30, 1))
```

## Exercise 3: Ward-round task planner


```python
def ward_round_task(patient_number):
    """
    Return a ward-round task label.

    Parameters
    ----------
    patient_number : int

    Returns
    -------
    str
    """
    if patient_number % 3 == 0 and patient_number % 5 == 0:
        return "blood pressure check + medication review"
    elif patient_number % 3 == 0:
        return "blood pressure check"
    elif patient_number % 5 == 0:
        return "medication review"
    else:
        return "routine review"
```

```python
# test cases
print(ward_round_task(1))
print(ward_round_task(2))
print(ward_round_task(3))
print(ward_round_task(5))
print(ward_round_task(10))
print(ward_round_task(15))
```

## Exercise 4: Calling a function inside a loop


```python
for patient_number in range(1, 16):
    task = ward_round_task(patient_number)
    print(f"Patient {patient_number}: {task}")
```

## Exercise 5: Looping and storing results


```python
tasks = []

for patient_number in range(1, 16):
    task = ward_round_task(patient_number)
    tasks.append(task)

print(tasks)
```

## Exercise 6: Oxygen saturation monitoring


```python
spo2 = 88
hour = 0
target_spo2 = 94

while spo2 < target_spo2:
    print(f"Hour {hour}: oxygen saturation is {spo2}%")
    spo2 += 2
    hour += 1

print(f"Target reached at hour {hour}: oxygen saturation is {spo2}%")
```

## Exercise 8: Create a simple alert list


```python
def heart_rate_in_normal_range(heart_rate_bpm):
    return heart_rate_bpm >= 60 and heart_rate_bpm <= 100
```

```python
patient_names = ["Asha", "Ben", "Chloe", "David"]
heart_rates = [72, 110, 58, 88]

alerts = []

for patient_name, heart_rate in zip(patient_names, heart_rates):
    if heart_rate_in_normal_range(heart_rate):
        alert = "normal"
    else:
        alert = "outside range"

    alerts.append(alert)
    print(f"{patient_name}: {alert}")

print(alerts)
```

## Exercise 10: Counting abnormal heart rates


```python
def heart_rate_in_normal_range(heart_rate_bpm):
    return heart_rate_bpm >= 60 and heart_rate_bpm <= 100
```

```python
def count_abnormal_heart_rates(heart_rates):
    count = 0

    for heart_rate in heart_rates:
        # notice the use of "not"
        if not heart_rate_in_normal_range(heart_rate):
            count += 1

    return count
```

```python
heart_rates = [72, 110, 58, 95, 102, 67]
print(count_abnormal_heart_rates(heart_rates))
```
