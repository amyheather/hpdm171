```python
# From Guided Example 1 — required by Exercise 2
def calculate_bmi(weight_kg, height_m):
    """
    Return body mass index (BMI) with no rounding

    Parameters
    ----------
    weight_kg : int or float
        Patient weight in kilograms.
    height_m : int or float
        Patient height in metres.

    Returns
    -------
    float
        Body mass index.

    Raises
    ------
    ValueError
        If weight_kg <= 0 or height_m <= 0.
    """
    if weight_kg <= 0:
        raise ValueError("weight_kg must be greater than 0")

    if height_m <= 0:
        raise ValueError("height_m must be greater than 0")

    bmi = weight_kg / (height_m ** 2)
    return bmi
```

## Exercise 1: Total drug dose


```python
def total_drug_dose(weight_kg, dose_mg_per_kg=5):
    """
    Return total drug dose in mg.
    """
    if weight_kg <= 0:
        raise ValueError("weight_kg must be greater than 0")
    if dose_mg_per_kg <= 0:
        raise ValueError("dose_mg_per_kg must be greater than 0")
    return weight_kg * dose_mg_per_kg

print(total_drug_dose(60))                    # expected: 300
print(total_drug_dose(72, dose_mg_per_kg=7))  # expected: 504
try:
    total_drug_dose(-5)
except ValueError as e:
    print(e)
```

## Exercise 2: BMI category from measurements


```python
def bmi_category_from_measurements(weight_kg, height_m):
    """
    Return a BMI category using weight and height.
    """
    bmi = calculate_bmi(weight_kg, height_m)
    if bmi < 18.5:
        return "underweight"
    elif bmi < 25:
        return "normal weight"
    elif bmi < 30:
        return "overweight"
    else:
        return "obese"

print(bmi_category_from_measurements(50, 1.70))   # underweight
print(bmi_category_from_measurements(70, 1.75))   # normal weight
print(bmi_category_from_measurements(85, 1.70))   # overweight
print(bmi_category_from_measurements(110, 1.70))  # obese
try:
    bmi_category_from_measurements(-10, 1.70)
except ValueError as e:
    print(e)
```

## Exercise 3: Observation range check


```python
def observation_outside_range(value, lower, upper):
    """
    Return True if a value is outside an inclusive range.
    """
    if lower > upper:
        return None
    return value < lower or value > upper

print(observation_outside_range(72, 60, 100))   # False
print(observation_outside_range(110, 60, 100))  # True
print(observation_outside_range(58, 60, 100))   # True
print(observation_outside_range(72, 100, 60))   # None
```

## Exercise 5: Update and query a patient record


```python
def add_observations(patient_record, observations):
    """
    Add a dictionary of observations to a patient record and return it.
    """
    patient_record.update(observations)
    return patient_record


def get_observation(patient_record, key, default="not recorded"):
    """
    Safely retrieve a value from a patient record.
    """
    return patient_record.get(key, default)


patient = make_patient_record(1003, 45, 88, ward="C1")
new_obs = {"heart_rate": 95, "oxygen_saturation": 98, "temperature": 37.2}
patient = add_observations(patient, new_obs)
print(patient)

print(get_observation(patient, "temperature"))     # expected: 37.2
print(get_observation(patient, "blood_pressure"))  # expected: not recorded
print(get_observation(patient, "ward"))            # expected: C1
```

```python
patient_record = {
    "patient_id": 1004,
    "demographics": {
        "age": 71,
        "ward": "A3"
    },
    "clinical_observations": {
        "heart_rate": 102,
        "oxygen_saturation": 94,
        "temperature": 38.6
    }
}

# Task 1: print the patient's ward
print(patient_record["demographics"]["ward"])
```

```python
# Task 2: update ward to "B1" and heart_rate to 96
patient_record["demographics"]["ward"] = "B1"
patient_record["clinical_observations"]["heart_rate"] = 96

print(patient_record)
```

```python
# Task 3:
def clinical_summary(patient_record):
    """
    Return a formatted string summary of clinical observations.

    Parameters
    ----------
    patient_record : dict
        Must contain a "clinical_observations" key whose value is a dict.

    Returns
    -------
    str
        Each observation on its own line in the format "key: value".
    """
    summary = ""
    for key, value in patient_record["clinical_observations"].items():
        summary += f"{key}: {value}\n"
    return summary.strip()


print(clinical_summary(patient_record))
```

## Exercise 8: Server Utilisation


```python
def server_utilisation(arrival_rate, service_rate):
    """
    Return server utilisation (rho) for an M/M/1 queue.

    Parameters
    ----------
    arrival_rate : float
        Mean arrival rate (lambda). Must be > 0.
    service_rate : float
        Mean service rate (mu). Must be > 0.

    Returns
    -------
    float
        Server utilisation rho = arrival_rate / service_rate.

    Raises
    ------
    ValueError
        If arrival_rate <= 0 or service_rate <= 0.
        If arrival_rate >= service_rate (unstable queue).
    """
    if arrival_rate <= 0:
        raise ValueError("arrival_rate must be greater than 0")
    if service_rate <= 0:
        raise ValueError("service_rate must be greater than 0")
    if arrival_rate >= service_rate:
        raise ValueError("Queue is unstable: arrival_rate must be less than service_rate")
    return arrival_rate / service_rate

print(round(server_utilisation(2.0, 3.0), 4))  # expected: 0.6667
print(round(server_utilisation(5.0, 8.0), 4))  # expected: 0.625
try:
    server_utilisation(3.0, 3.0)
except ValueError as e:
    print(e)
```

## Exercise 9: M/M/1 waiting time


```python
def waiting_time(arrival_rate, service_rate):
    """
    Return average waiting time in queue for an M/M/1 model.
    """
    return arrival_rate / (service_rate * (service_rate - arrival_rate))


def prob_n_customers_in_system(n, arrival_rate, service_rate):
    """Prob n customers are in an M/M/1 system."""
    rho = server_utilisation(arrival_rate, service_rate)
    return (1 - rho) * (rho ** n)


arrival_rate = 2.0
service_rate = 3.0

print(round(waiting_time(arrival_rate, service_rate), 4))                   # expected: 0.6667
print(round(prob_n_customers_in_system(0, arrival_rate, service_rate), 4))  # expected: 0.3333
print(round(prob_n_customers_in_system(1, arrival_rate, service_rate), 4))  # expected: 0.2222
print(round(prob_n_customers_in_system(2, arrival_rate, service_rate), 4))  # expected: 0.1481
```

## Exercise 10: Analysing a Set of Queues with Dictionaries


```python
def summarise_clinic(clinic):
    """
    Return a summary dictionary for one M/M/1 clinic.

    Parameters
    ----------
    clinic : dict
        Must contain keys 'name', 'arrival_rate', 'service_rate'.

    Returns
    -------
    dict
        Keys: 'name', 'utilisation', 'mean_wait_hours'.

    Raises
    ------
    ValueError
        Propagated from server_utilisation() if the queue is unstable.
    """
    rho = server_utilisation(
        arrival_rate=clinic["arrival_rate"],
        service_rate=clinic["service_rate"]
    )
    wq = waiting_time(
        arrival_rate=clinic["arrival_rate"],
        service_rate=clinic["service_rate"]
    )
    return {
        "name": clinic["name"],
        "utilisation": round(rho, 3),
        "mean_wait_hours": round(wq, 3)
    }


def find_busiest_clinic(clinics):
    """
    Return the name of the clinic with the highest server utilisation.

    Parameters
    ----------
    clinics : list of dict
        Each dict must contain 'name', 'arrival_rate', 'service_rate'.

    Returns
    -------
    str
        Name of the busiest clinic.
    """
    summaries = [summarise_clinic(clinic) for clinic in clinics]
    busiest = max(summaries, key=lambda r: r["utilisation"])
    return busiest["name"]


clinics = [
    {"name": "Cardiology", "arrival_rate": 4.0, "service_rate": 6.0},
    {"name": "Dermatology", "arrival_rate": 3.0, "service_rate": 4.0},
    {"name": "Orthopaedics", "arrival_rate": 5.0, "service_rate": 5.5},
    {"name": "Ophthalmology", "arrival_rate": 2.5, "service_rate": 3.0},
]

# Task 1
for clinic in clinics:
    print(summarise_clinic(clinic))

# Task 2
print(find_busiest_clinic(clinics))  # expected: Orthopaedics
```
