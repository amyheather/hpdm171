## 1. Why functions matter

Functions help us write code once and reuse it many times.
This is especially useful when we need to repeat the same calculation for many patients.


```python
weight_kg = 82
height_m = 1.78
bmi = weight_kg / (height_m ** 2)
print(round(bmi, 1))
```

That works for one patient, but what if we need to repeat the same calculation many times?
A function lets us package the logic into a reusable block of code.


```python
def calculate_bmi(weight_kg, height_m):
    """Return body mass index (BMI) for one patient."""
    return weight_kg / (height_m ** 2)
```

```python
patient_bmi = calculate_bmi(82, 1.78)
print(round(patient_bmi, 1))
```

```python
patient_bmi = calculate_bmi(70, 1.80)
print(round(patient_bmi, 1))
```

### 1.1. A quick reminder on docstrings

A **docstring** is a short description written inside a function.

We often use docstrings in real Python code so that `help()` can explain what a function does.

> In this lecture I will mostly leave docstrings out or keep them minimal, to save space on screen, but it is useful to see one example.

> Make sure you include docstrings in real code and in coursework.

```python
def calculate_bmi(weight_kg, height_m):
    """
    Calculate body mass index (BMI) for one patient.

    Parameters
    ----------
    weight_kg : int or float
        Patient weight in kilograms.
    height_m : int or float
        Patient height in metres.

    Returns
    -------
    float
        Body mass index calculated as weight divided by height squared.
    """
    return weight_kg / (height_m ** 2)
```

### 1.2 Scientific coding habits

When we write Python for scientific or health-data work, we do not just want code that works once.

We usually want code that is:

- reusable
- readable
- easy to test
- clear when inputs are impossible

A common Python approach is to **raise an exception** when the inputs to a function do not make sense.

For example, a patient's weight cannot be negative, and height cannot be zero.

```python
def calculate_bmi(weight_kg, height_m):
    """
    Return body mass index (BMI) with no rounding

    Parameters
    ----------
    weight_kg : int or float
        Patient weight in kilograms.
    height_m : int or float
        Patient height in metres.

    Returns
    -------
    float
        Body mass index.

    Raises
    ------
    ValueError
        If weight_kg <= 0 or height_m <= 0.
    """
    if weight_kg <= 0:
        raise ValueError("weight_kg must be greater than 0")

    if height_m <= 0:
        raise ValueError("height_m must be greater than 0")

    bmi = weight_kg / (height_m ** 2)
    return bmi
```

```python
print(calculate_bmi(82, 1.78))
print(calculate_bmi(70, 1.80))
```

This function is still small, but it already shows several useful scientific coding ideas:

1. all the information needed by the function is passed in as parameters
2. the function returns a value instead of printing from inside the function
3. the parameter names are descriptive
4. the docstring explains what the function does
5. the function checks for impossible (invalid) inputs

### 1.3. A note on invalid inputs

In Python, a common way to report invalid inputs is to **raise an exception**.

Here we use `ValueError` to say that the function was given a value that does not make sense, such as a negative weight or zero height.

At this stage, you do not need to learn full exception handling in detail.

For now, the key idea is:

- valid input -> return a useful result
- invalid input -> stop clearly and report the problem

```python
print(calculate_bmi(-1, 1.78))
```

### 1.4 Testing more than one case

A scientific function should be tested on more than one input.

A simple starting point is to test:

- a normal case
- another sensible case
- an invalid case

```python
# weight, height, expected answer (1dp)
test_cases = [
    (82, 1.78, 25.9),
    (70, 1.80, 21.6),
]

for weight_kg, height_m, expected in test_cases:
    result = calculate_bmi(weight_kg, height_m)
    print(
        f"weight_kg={weight_kg}, height_m={height_m}, "
        f"expected={expected}, got={round(result, 1)}"
    )
```

## 2. Parameters and return values

Functions usually take inputs called parameters and send back an output using `return`.


```python
def classify_age(age):
    """Return 'child' if `age` under 18 else 'adult'"""
    if age < 18:
        return "child"
    else:
        return "adult"
```

```python
# we pass the integer 12 as the `age` parameters
# and assign the return value to `classification_1`
classification_1 = classify_age(12)
print(classification_1)

patient_age = 45
classification_2 = classify_age(patient_age)
print(classification_2)
```

## 3. Scope in notebooks and functions

Scope means the part of a program where a variable can be accessed.
In Jupyter notebooks, it is helpful to think about two common kinds of scope:

* **Notebook-level scope**: variables created in notebook cells and available later if those cells have been run.
* **Function-level scope**: variables created inside a function and normally only available inside that function.


### 3.1 Notebook-level variables (and why to avoid them in functions)

A function *can* read a variable that already exists at the notebook level (a global variable). 

However, **this is considered bad practice**. If your function relies on variables outside of itself, it becomes very difficult to reuse in other notebooks, and it can cause hidden bugs if you run your notebook cells out of order.

```python
# This is a notebook level variable it has "global" scope
clinic_name = "Exeter Community Clinic"
```

```python
def attendance_message(patient_name):
    # BAD PRACTICE: Relying on a notebook-level variable `clinic_name`
    return f"{patient_name} attended {clinic_name}"
```

```python
print(attendance_message("Amber"))
print(clinic_name)
```

**Best Practice: Pass `clinic_name` as a parameter!**

Instead of letting the function look outside for information, we should pass all the information the function needs directly through its parameters. 

This makes the function completely self-contained, predictable, and reusable.

```python
def attendance_message_good(patient_name, clinic):
    # This function only uses variables passed into it.
    return f"{patient_name} attended {clinic}"
```

```python
current_clinic = "Exeter Community Clinic"

# We pass the notebook-level variable into the function explicitly
print(attendance_message_good("Amber", current_clinic))
print(attendance_message_good("David", "Royal Devon Hospital"))
```

### 3.2 Local variables stay inside the function

Variables created inside a function are usually local variables.
We cannot normally use them outside the function.


```python
def calculate_square(number):
    squared_value = number ** 2
    return squared_value
```

```python
print(calculate_square(4))
```

```python
# this will generate a Python NameError as the variable is not in scope
print(squared_value)
```

### 3.3 A common source of confusion

If we assign to a variable inside a function, Python treats it as local unless we explicitly say otherwise.
That means changing a variable name inside a function does not usually change the notebook-level variable with the same name.


```python
# this is a notebook level variable with global scope
status = "not reviewed"
```

```python
def review_patient():
    # this local variable has the same name as the variable with global scope
    status = "reviewed"
    print("Inside function:", status)
```

```python
print("Before function call:", status)
review_patient()
print("After function call:", status)
```

In the previous example, there are two different variables called `status`.
One lives at notebook level and one lives inside the function.
Using the same variable name in two scopes can be confusing for beginners, so clear names help.

## 4. Optional arguments

Functions can have default values for parameters.
These are useful when there is a common case. 
When parameters have a default value they become **optional**.


```python
def calculate_bmi(weight_kg, height_m, decimals=1):
    bmi = weight_kg / (height_m ** 2)
    return round(bmi, decimals)
```

```python
print(calculate_bmi(82, 1.78))
print(calculate_bmi(82, 1.78, 2))
print(calculate_bmi(weight_kg=82, height_m=1.78, decimals=3))
```

```python
def high_heart_rate(heart_rate, threshold=100):
    return heart_rate > threshold
```

```python
print(high_heart_rate(110))
print(high_heart_rate(110, threshold=120))
```

## 5. Returning multiple values

A Python function can return more than one value.
The returned values come back as a tuple.


```python
def summarise_wait_times(wait_times):
    minimum = min(wait_times)
    maximum = max(wait_times)
    average = sum(wait_times) / len(wait_times)
    return minimum, maximum, average
```

```python
summary = summarise_wait_times([18, 25, 42, 31, 20])
print(summary)
print(type(summary))
```

We can **unpack** the `tuple` returned into separate variables.

```python
min_wait, max_wait, mean_wait = summarise_wait_times([18, 25, 42, 31, 20])
```

```python
print(min_wait)
print(max_wait)
print(round(mean_wait, 1))
```

## 7. Dictionaries

A dictionary stores data as key-value pairs.
This is useful when one patient has several pieces of information.


```python
patient = {
    "id": 1001,
    "age": 67,
    "heart_rate": 88,
    "ward": "A3"
}
```

```python
print(patient)
```

```python
print(patient["age"])
print(patient["ward"])
```

### 7.1 Updating one item at a time

We can change a dictionary by assigning a new value to an existing key.

If the key does not exist yet, Python adds it to the dictionary.

```python
patient = {
    "id": 1001,
    "age": 67,
    "heart_rate": 88,
    "ward": "A3"
}

# overwrite existing heart rate
patient["heart_rate"] = 92

# add a new key-value pair to the `dict`
patient["oxygen_saturation"] = 98

print(patient)
```

### 7.2 Updating with another dictionary

Sometimes we receive several new pieces of information at once.

The `update()` method lets us use one dictionary to update another.

```python
patient = {
    "id": 1001,
    "age": 67,
    "heart_rate": 88,
    "ward": "A3"
}

new_observations = {
    "heart_rate": 92,
    "oxygen_saturation": 98,
    "temperature": 37.4
}

patient.update(new_observations)

print(patient)
```

### 7.3 Nested Dictionaries

In real-world health data, information is often grouped together. We can store dictionaries inside other dictionaries. This is called a **nested dictionary**.

For example, a patient might have general details, but also a specific group of clinical observations.

```python
patient_record = {
    "patient_id": 1001,
    # demographics is also a dict. It is "nested"
    "demographics": {
        "age": 67,
        "ward": "A3"
    },
    # another nested dict
    "clinical_observations": {
        "heart_rate": 88,
        "blood_pressure": "120/80"
    }
}

print("Full record:")
print(patient_record)
```

To access data inside a nested dictionary, we just chain the square brackets together, following the path to the data we want.

```python
# Accessing the outer dictionary
print("Demographics:")
print(patient_record["demographics"])

# Chaining brackets to access the inner dictionary
print("\nSpecific age:")
print(patient_record["demographics"]["age"])

print("\nSpecific heart rate:")
print(patient_record["clinical_observations"]["heart_rate"])
```

You can update nested dictionaries the exact same way we updated standard dictionaries.

```python
# The patient's heart rate changes
patient_record["clinical_observations"]["heart_rate"] = 95

# The patient is moved to a new ward
patient_record["demographics"]["ward"] = "B2"

print("Updated clinical observations:")
print(patient_record["clinical_observations"])
```

### 7.4 Iterating over key-value pairs with `.items()`

Sometimes we want to look at every piece of information stored in a dictionary.

The `.items()` method lets us loop through both the **key** and the **value** at the same time.

This is useful when we want to print a patient record, check all fields, or summarise the contents of a dictionary.

```python
patient = {
    "id": 1001,
    "age": 67,
    "heart_rate": 88,
    "ward": "A3"
}
```

```python
# iterate over key value pairs
for key, value in patient.items():
    print(key, value)
```

### 7.5 Safe access with `get()`

Sometimes a dictionary might not contain the key we ask for. Using the access methods we have seen so far Python will raise a `KeyError` and crash!

Using `dict.get()` is a safe way to look up a value.

If the key exists, it returns the value.
If the key does not exist, it returns `None` by default, or a value we choose.

```python
patient = {
    "id": 1001,
    "age": 67,
    "heart_rate": 88,
    "ward": "A3"
}

print(patient.get("ward"))
print(patient.get("oxygen_saturation"))
print(patient.get("oxygen_saturation", "Not recorded"))
```

```python
print(patient.get("temperature"))       # what does this return?
```

```python
print(patient["temperature"])           # what does this return?
```

## 8. Functions with dictionaries as parameters

Now we combine functions, conditionals, loops, and dictionaries.


```python
def patient_is_high_risk(patient_record, heart_rate_threshold=100):
    return patient_record["age"] >= 65 and patient_record["heart_rate"] > heart_rate_threshold
```

```python
patient_1 = {"id": 1, "age": 72, "heart_rate": 108}
patient_2 = {"id": 2, "age": 40, "heart_rate": 88}

print(patient_is_high_risk(patient_1))
print(patient_is_high_risk(patient_2))
```

```python
patients = [
    {"id": 1, "age": 72, "heart_rate": 108},
    {"id": 2, "age": 40, "heart_rate": 88},
    {"id": 3, "age": 69, "heart_rate": 95},
    {"id": 4, "age": 83, "heart_rate": 115}
]

for patient in patients:
    if patient_is_high_risk(patient):
        print(f"Patient {patient['id']} is high risk")
```

```python
def count_high_risk_patients(patient_records, heart_rate_threshold=100):
    """
    Count the number of high-risk patients in a list of patient records.

    Parameters
    ----------
    patient_records : list of dict
        A list where each item is a patient record dictionary.
    heart_rate_threshold : int or float, optional
        The heart rate threshold above which a patient is considered
        high risk. The default is 100.

    Returns
    -------
    int
        The number of patients classified as high risk.
    """
    count = 0

    for patient in patient_records:
        if patient_is_high_risk(patient, heart_rate_threshold):
            count += 1

    return count
```

```python
patients = [
    {"id": 1, "age": 72, "heart_rate": 108},
    {"id": 2, "age": 40, "heart_rate": 88},
    {"id": 3, "age": 69, "heart_rate": 95},
    {"id": 4, "age": 83, "heart_rate": 115}
]

print(count_high_risk_patients(patients))
```
