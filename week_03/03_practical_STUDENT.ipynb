## Guided example 1: BMI as a reusable function

A useful scientific function should:
- accept all required information as parameters
- return a result
- use clear variable names
- include a short docstring
- handle impossible inputs in a simple way

In this notebook, if the inputs are not valid, the function will return `None`.


````python
def calculate_bmi(weight_kg, height_m):
    """
    Return body mass index (BMI) with no rounding

    Parameters
    ----------
    weight_kg : int or float
        Patient weight in kilograms.
    height_m : int or float
        Patient height in metres.

    Returns
    -------
    float
        Body mass index.

    Raises
    ------
    ValueError
        If weight_kg <= 0 or height_m <= 0.
    """
    if weight_kg <= 0:
        raise ValueError("weight_kg must be greater than 0")

    if height_m <= 0:
        raise ValueError("height_m must be greater than 0")

    bmi = weight_kg / (height_m ** 2)
    return bmi
````

````python
print(calculate_bmi(82, 1.78))
print(calculate_bmi(70, 1.80))
````

````python
# this will raise a value error!
print(calculate_bmi(-70, 1.80)) 
````

## Guided example 2: Returning more than one value

Sometimes one function should return several summary values. 

````python
def summarise_wait_times(wait_times):
    """
    Return the minimum, maximum, and mean of a list of wait times.
    If the list is empty, return None.
    """
    if len(wait_times) == 0:
        raise ValueError("list should contained at least 1 item")

    minimum = min(wait_times)
    maximum = max(wait_times)
    mean_wait = sum(wait_times) / len(wait_times)

    # this returns a tuple
    return minimum, maximum, mean_wait
````

````python
summary = summarise_wait_times([18, 25, 42, 31, 20])
print(summary)
````

````python
# tuple unpacking
min_wait, max_wait, mean_wait = summarise_wait_times([18, 25, 42, 31, 20])
````

## Exercise 8: Server Utilisation

In any health service system, e.g., a clinic, an A&E department, a pharmacy, a key question
is: how busy is the server (the nurse, the doctor, the machine)?

**Server utilisation** (ρ, "rho") answers this. It is the fraction of time the server is busy, defined as:

$$\rho = \frac{\lambda}{\mu}$$

where $\lambda$ is the arrival rate and $\mu$ is the service rate.

If $\rho \geq 1$, the server can never keep up with demand and the queue grows
without bound i.e. it just keeps getting longer!  In mathematical terms we say that the system is **unstable**.

## Exercise 9: Scientific coding challenge : M/M/1 waiting time

Health services like A&E departments and outpatient clinics can often by modelled as queuing systems. Where patients wait in a "queue" for a medical resource like a nurse, doctor or scanner.

A classic **M**arkovian mathematical model for representing the steady state performance of queuing and service systems is the **M**/**M**/1.  

In the Markovian model "customers" arrive to the model following a Poisson process with rate parameter $\lambda$ (e.g. an average of 5 arrivals per hour). The M/M/1 is a first come first served system with a single server. Service time is exponentially distributed with mean $1/\mu$ i.e. customers are served at rate $\mu$.

The equation for calculating the average waiting time to be served ($W_q$) in an M/M/1 queue is:

$$W_q = \dfrac{\lambda}{\mu\left(\mu-\lambda\right)}$$

While the probability that there are $n$ customers in the system ($P_n$) is given by:

$$P_n = \left(1 - \dfrac{\lambda}{\mu}\right)\left(\dfrac{\lambda}{\mu}\right)^n$$


**Example:**

Given:

$$\lambda = 2 \textnormal{ per hour}$$
$$\mu = 3 \textnormal{ per hour}$$

Calculate $W_q$ and $P_0, P_1$

$$W_q = \dfrac{2}{3\left(3 - 2\right)} = \dfrac{2}{3} \textnormal{ per hour}$$


$$P_0 =  \left(1 - \dfrac{\lambda}{\mu}\right) = 1 - \dfrac{2}{3} = \dfrac{1}{3}$$

$$P_1 =  \left(\dfrac{1}{3}\right)\left(\dfrac{2}{3}\right) = \dfrac{2}{9}$$
